### Training functions

Firstly, we use the *train_net* function defined in *train.py* to obtain a model with specific features. The function takes the following arguments:

- *train_data*: one of 'Random', 'MNIST' or 'EMNIST', specifying the dataset used to train the model.
- *network*: 'CNN' or 'Fully', depending on the network to be trained.
- *save_vars*: a boolean indicating whether model variables should be saved.
- *test_data* (optional): one of 'Random', 'MNIST' or 'EMNIST', specifying the dataset used to test the model. If no option is specified, it will use the same as *train_data*.
  
We execute the function for each possible combination of model settings.

```python
from XORclassic.train import train_net

for network in ['CNN', 'Fully']:
    for train_data in ['Random', 'MNIST', 'EMNIST']:
        train_net(train_data = train_data, network = network, save_vars = True)
```

Then, we use the *train_class* function defined in the same file than the previous one, to obtain the linear classifier for a specific reconstruction model. The function takes the following arguments:

- *classif_data*: one of 'Random', 'MNIST' or 'EMNIST', specifying the dataset used to train and test the model.
- *train_data* and *network*: same as before; these are used to load the previously saved reconstruction model.
- *save_vars*: a boolean indicating whether model variables should be saved.

We execute the function for each possible combination of model settings.

```python
from XORclassic.train import train_class

for network in ['CNN', 'Fully']:
    for train_data in ['Random', 'MNIST', 'EMNIST']:
        for classif_data in['MNIST', 'EMNIST']:
            train_class(classif_data = classif_data, train_data = train_data, network = network, save_vars = True)
```

### Functions to analyse
